## Predict a change

Will `alias` have three items after we append through `readings`? Will `separate`?
Write a prediction before running the following cell.


```python
readings = [False, True]
alias = readings
separate = readings.copy()
readings.append(False)
print("Original:", readings)
print("Alias:", alias)
print("Separate copy:", separate)
```

`alias` refers to the same list, so it sees the appended item. `copy()` makes a
new list. Here that is enough because the items are booleans; this is a shallow
copy, so nested mutable objects would need more care.

## Notice cell state

Run the next cell twice. Predict both lengths. The second run uses the state left
by the first run, even if you didn't change any code.


```python
readings.append(True)
print("Length now:", len(readings))
```

From a fresh top-to-bottom run, the length is 4. Running only the last cell
again makes it 5. Rerun the first code cell to recreate the lists, or restart the
kernel and run all. This is why a result from an old notebook state can be misleading.
